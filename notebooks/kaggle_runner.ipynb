# OPTIMIZER AUTOPSY — Kaggle runner
Thin launcher. Sets `CUBLAS_WORKSPACE_CONFIG` **before** torch is imported, clones the repo, loads secrets, exposes `run(cmd)`. See `research/README.md`.

```python
# MUST run before any torch import (determinism).
import os
os.environ["CUBLAS_WORKSPACE_CONFIG"] = ":4096:8"
```

```python
# Clone the repo into the Kaggle working dir.
REPO = "https://github.com/jaztulsi/optimizer-autopsy.git"
import os, subprocess
if not os.path.isdir("optimizer-autopsy"):
    subprocess.run(["git", "clone", REPO], check=True)
os.chdir("optimizer-autopsy")
subprocess.run(["pip", "install", "-q", "-r", "research/requirements.txt"], check=True)
```

```python
# Secrets from Kaggle Secrets -> env, so research.harness.secrets can resolve them.
from kaggle_secrets import UserSecretsClient
_s = UserSecretsClient()
for k in ("HF_TOKEN", "WANDB_API_KEY"):
    try:
        os.environ[k] = _s.get_secret(k)
    except Exception as e:
        print(f"no secret {k}: {e}")
```

```python
import subprocess
def run(cmd):
    """Run a shell command (e.g. 'python -m research.experiments.proxy.smoke') streaming output."""
    return subprocess.run(cmd, shell=True, check=True)

# run("python -m research.experiments.proxy.smoke")
```
